```python
import pandas as pd
import sklearn.datasets
import matplotlib as mt
```

```python
df = sklearn.datasets.load_breast_cancer()
dfm = pd.DataFrame(df.data, columns=df.feature_names)
dfm.head()
```

```python
x=dfm['worst radius']
#print(x)
x=x.head()
y=dfm['worst perimeter']
y=y.head()
#print(y)
x_mean=x.mean()
y_mean=y.mean()
w1=((x-x_mean)*(y-y_mean)).sum()/((x-x_mean)**2).sum()
w0=y_mean-w1*x_mean
y=w0+w1*x
import matplotlib.pyplot as plt
plt.plot(x,y,'o',linestyle ='-')
plt.show()
```
